```python
!pip install -q tensorflow==2.15.0 protobuf==3.20.* >/dev/null

import os, cv2, random, shutil, glob, json, math, itertools
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns
import pandas as pd

#Config & Paths
random.seed(42)
np.random.seed(42)
tf.random.set_seed(42)

print("TensorFlow:", tf.__version__)

DATA_DIR = "/kaggle/input/the-iqothnccd-lung-cancer-dataset/The IQ-OTHNCCD lung cancer dataset"
INVALID_DIR = "/kaggle/input/invalid-class-dataset/Invalid Class"
OUTPUT_DIR = "/kaggle/working/augmented_dataset"
split_dir = "/kaggle/working/splitted_dataset"
RESULTS_DIR = "/kaggle/working/results"
TFLITE_DIR = os.path.join(RESULTS_DIR, "tflite_models")
os.makedirs(OUTPUT_DIR, exist_ok=True)
os.makedirs(RESULTS_DIR, exist_ok=True)
os.makedirs(TFLITE_DIR, exist_ok=True)
os.makedirs(split_dir, exist_ok=True)

classes = ["Normal cases", "Bengin cases", "Malignant cases", "Invalid Class"]
for cls in classes:
    os.makedirs(os.path.join(OUTPUT_DIR, cls), exist_ok=True)
    for s in ["train","val","test"]:
        os.makedirs(os.path.join(split_dir, s, cls), exist_ok=True)

IMG_SIZE = (224, 224)
SEG_SIZE = (128, 128)
BATCH_SIZE = 32
TARGET_COUNT = 10000

#Augmentation
datagen = ImageDataGenerator(
    rotation_range=25,
    width_shift_range=0.15,
    height_shift_range=0.15,
    shear_range=0.15,
    zoom_range=0.15,
    horizontal_flip=True,
    vertical_flip=True,
    fill_mode='nearest',
    brightness_range=[0.8, 1.2]
)

def augment_class(cls, class_path):
    images = [f for f in os.listdir(class_path) if f.lower().endswith(('.png','.jpg','.jpeg'))]
    print(f"{cls}: Original count = {len(images)}")
    existing = len(os.listdir(os.path.join(OUTPUT_DIR, cls)))
    if existing >= TARGET_COUNT:
        print(f"Skipping {cls}, already has {existing} images.")
        return
    count = 0
    # Copy originals
    for img_name in images:
        src = os.path.join(class_path, img_name)
        dst = os.path.join(OUTPUT_DIR, cls, f"org_{img_name}")
        try:
            shutil.copy(src, dst)
            count += 1
        except:
            pass
    # Augment until target
    while count < TARGET_COUNT and len(images) > 0:
        for img_name in images:
            img_path = os.path.join(class_path, img_name)
            img = cv2.imread(img_path)
            if img is None: continue
            img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
            img = cv2.resize(img, IMG_SIZE)
            x = np.expand_dims(img, 0)
            aug_iter = datagen.flow(x, batch_size=1)
            # create up to 3 augmented per source each loop
            for _ in range(3):
                aug_img = next(aug_iter)[0].astype(np.uint8)
                save_path = os.path.join(OUTPUT_DIR, cls, f"{cls}_aug_{count}.jpg")
                cv2.imwrite(save_path, cv2.cvtColor(aug_img, cv2.COLOR_RGB2BGR))
                count += 1
                if count >= TARGET_COUNT: break
            if count >= TARGET_COUNT: break
    print(f"{cls}: Final count = {len(os.listdir(os.path.join(OUTPUT_DIR, cls)))}")

for cls in classes:
    class_path = INVALID_DIR if cls == "Invalid Class" else os.path.join(DATA_DIR, cls)
    if os.path.exists(class_path):
        augment_class(cls, class_path)
    else:
        print("Warning: path missing for", cls, class_path)

#Split (train/val/test)
if not os.listdir(os.path.join(split_dir, "train", "Normal cases")):
    train_ratio, val_ratio, test_ratio = 0.7, 0.15, 0.15
    for cls in classes:
        cls_path = os.path.join(OUTPUT_DIR, cls)
        images = [f for f in os.listdir(cls_path) if f.endswith(".jpg") or f.startswith("org_")]
        random.shuffle(images)
        n_total = len(images)
        n_train = int(n_total * train_ratio)
        n_val = int(n_total * val_ratio)
        for img in images[:n_train]:
            shutil.copy(os.path.join(cls_path, img), os.path.join(split_dir, "train", cls, img))
        for img in images[n_train:n_train+n_val]:
            shutil.copy(os.path.join(cls_path, img), os.path.join(split_dir, "val", cls, img))
        for img in images[n_train+n_val:]:
            shutil.copy(os.path.join(cls_path, img), os.path.join(split_dir, "test", cls, img))
    print("Splitting Done.")

#Data Generators
train_datagen = ImageDataGenerator(preprocessing_function=preprocess_input,
                                   rotation_range=20, width_shift_range=0.1, height_shift_range=0.1,
                                   zoom_range=0.1, horizontal_flip=True, fill_mode='nearest')

val_datagen = ImageDataGenerator(preprocessing_function=preprocess_input)
test_datagen = ImageDataGenerator(preprocessing_function=preprocess_input)

TRAIN_DIR = os.path.join(split_dir, "train")
VAL_DIR = os.path.join(split_dir, "val")
TEST_DIR = os.path.join(split_dir, "test")

train_gen = train_datagen.flow_from_directory(TRAIN_DIR, target_size=IMG_SIZE, batch_size=BATCH_SIZE, class_mode="categorical", shuffle=True)
val_gen = val_datagen.flow_from_directory(VAL_DIR, target_size=IMG_SIZE, batch_size=BATCH_SIZE, class_mode="categorical", shuffle=False)
test_gen = test_datagen.flow_from_directory(TEST_DIR, target_size=IMG_SIZE, batch_size=1, class_mode="categorical", shuffle=False)

# compute class weights
labels_for_weight = train_gen.classes
class_weights_list = compute_class_weight('balanced', classes=np.unique(labels_for_weight), y=labels_for_weight)
class_weights = {i: float(w) for i, w in enumerate(class_weights_list)}
print("Class weights:", class_weights)

#Classification Model (MobileNetV2)
base_model = tf.keras.applications.MobileNetV2(input_shape=(IMG_SIZE[0], IMG_SIZE[1], 3), include_top=False, weights="imagenet")
base_model.trainable = False

inputs = layers.Input(shape=(IMG_SIZE[0], IMG_SIZE[1], 3))
x = base_model(inputs, training=False)
x = layers.GlobalAveragePooling2D()(x)
x = layers.BatchNormalization()(x)
x = layers.Dense(256, activation="relu")(x)
x = layers.Dropout(0.4)(x)
outputs = layers.Dense(len(train_gen.class_indices), activation="softmax")(x)

clf_model = models.Model(inputs, outputs)
clf_model.compile(optimizer=tf.keras.optimizers.Adam(1e-3), loss="categorical_crossentropy", metrics=["accuracy"])

callbacks = [
    tf.keras.callbacks.EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True, verbose=1),
    tf.keras.callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.2, patience=2, verbose=1),
    tf.keras.callbacks.ModelCheckpoint(os.path.join(RESULTS_DIR, "best_clf.h5"), monitor='val_loss', save_best_only=True, verbose=1)
]

print("Starting Initial Training (classification)...")
history_clf = clf_model.fit(train_gen, validation_data=val_gen, epochs=6, callbacks=callbacks, class_weight=class_weights)

# Fine-tuning (unfreeze top layers)
print("Starting Fine-tuning (classification)...")
base_model.trainable = True
for layer in base_model.layers[:-80]:
    layer.trainable = False

clf_model.compile(optimizer=tf.keras.optimizers.Adam(1e-5), loss="categorical_crossentropy", metrics=["accuracy"])
history_finetune = clf_model.fit(train_gen, validation_data=val_gen, epochs=8, callbacks=callbacks, class_weight=class_weights)

clf_model.save(os.path.join(RESULTS_DIR, "final_classification_model.h5"))

# Plot classification training history
def plot_history(h1, h2=None, title_prefix="Classification"):
    plt.figure(figsize=(14,5))
    # loss
    plt.subplot(1,2,1)
    plt.plot(h1.history['loss'], label='train_loss_init')
    plt.plot(h1.history['val_loss'], label='val_loss_init')
    if h2:
        plt.plot(h2.history['loss'], label='train_loss_ft')
        plt.plot(h2.history['val_loss'], label='val_loss_ft')
    plt.title(f'{title_prefix} Loss')
    plt.legend()
    # acc
    plt.subplot(1,2,2)
    plt.plot(h1.history.get('accuracy',[]), label='train_acc_init')
    plt.plot(h1.history.get('val_accuracy',[]), label='val_acc_init')
    if h2:
        plt.plot(h2.history.get('accuracy',[]), label='train_acc_ft')
        plt.plot(h2.history.get('val_accuracy',[]), label='val_acc_ft')
    plt.title(f'{title_prefix} Accuracy')
    plt.legend()
    plt.show()

plot_history(history_clf, history_finetune)

#Segmentation Data Prep & U-Net
def preprocess_seg_input(path, target_size=SEG_SIZE):
    img = cv2.imread(path)
    if img is None: return None
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    img = cv2.resize(img, target_size)
    return img.astype(np.float32) / 255.0

def create_mask(img, target_size=SEG_SIZE):
    # img assumed 0-1 float RGB
    gray = cv2.cvtColor((img * 255).astype(np.uint8), cv2.COLOR_RGB2GRAY)
    # CLAHE to enhance contrast (helps Otsu)
    clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8))
    gray_clahe = clahe.apply(gray)
    gray_clahe = cv2.resize(gray_clahe, target_size)
    _, mask = cv2.threshold(gray_clahe, 0, 1, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    kernel = np.ones((3, 3), np.uint8)
    mask = cv2.morphologyEx(mask.astype(np.uint8), cv2.MORPH_CLOSE, kernel)
    mask = cv2.morphologyEx(mask, cv2.MORPH_OPEN, kernel)
    return mask[..., np.newaxis].astype(np.float32)

seg_classes = ["Bengin cases", "Malignant cases"]
X_seg, Y_seg = [], []
for cls in seg_classes:
    files = glob.glob(os.path.join(TRAIN_DIR, cls, "org_*"))
    if len(files) < 100:
        files = glob.glob(os.path.join(TRAIN_DIR, cls, "*"))
    for f in files[:800]:  # cap to avoid memory issues
        img = preprocess_seg_input(f, SEG_SIZE)
        if img is None: continue
        mask = create_mask(img, SEG_SIZE)
        X_seg.append(img)
        Y_seg.append(mask)

X_seg, Y_seg = np.array(X_seg), np.array(Y_seg)
print("Segmentation data shapes:", X_seg.shape, Y_seg.shape)

def build_unet(input_shape=(SEG_SIZE[0], SEG_SIZE[1], 3)):
    inputs = layers.Input(input_shape)
    c1 = layers.Conv2D(32, 3, activation="relu", padding="same")(inputs)
    c1 = layers.Conv2D(32, 3, activation="relu", padding="same")(c1)
    p1 = layers.MaxPool2D()(c1)
    c2 = layers.Conv2D(64, 3, activation="relu", padding="same")(p1)
    c2 = layers.Conv2D(64, 3, activation="relu", padding="same")(c2)
    p2 = layers.MaxPool2D()(c2)
    c3 = layers.Conv2D(128, 3, activation="relu", padding="same")(p2)
    c3 = layers.Conv2D(128, 3, activation="relu", padding="same")(c3)
    u2 = layers.Conv2DTranspose(64, 2, strides=2, padding="same")(c3)
    u2 = layers.Concatenate()([u2, c2])
    c4 = layers.Conv2D(64, 3, activation="relu", padding="same")(u2)
    c4 = layers.Conv2D(64, 3, activation="relu", padding="same")(c4)
    u1 = layers.Conv2DTranspose(32, 2, strides=2, padding="same")(c4)
    u1 = layers.Concatenate()([u1, c1])
    c5 = layers.Conv2D(32, 3, activation="relu", padding="same")(u1)
    c5 = layers.Conv2D(32, 3, activation="relu", padding="same")(c5)
    outputs = layers.Conv2D(1, 1, activation="sigmoid")(c5)
    return models.Model(inputs, outputs)

seg_model = None
history_seg = None
if len(X_seg) > 50:
    seg_model = build_unet((SEG_SIZE[0], SEG_SIZE[1], 3))
    seg_model.compile(optimizer=tf.keras.optimizers.Adam(1e-4), loss="binary_crossentropy", metrics=["accuracy"])
    seg_callbacks = [
        tf.keras.callbacks.EarlyStopping(monitor='val_loss', patience=6, restore_best_weights=True, verbose=1),
        tf.keras.callbacks.ModelCheckpoint(os.path.join(RESULTS_DIR, "best_seg.h5"), save_best_only=True, verbose=1)
    ]
    history_seg = seg_model.fit(X_seg, Y_seg, batch_size=16, epochs=25, validation_split=0.15, callbacks=seg_callbacks)
    seg_model.save(os.path.join(RESULTS_DIR, "final_segmentation_model.h5"))
else:
    print("Not enough data for segmentation training. Skipping segmentation training.")

# Plot segmentation history if present
if history_seg is not None:
    plt.figure(figsize=(12,4))
    plt.subplot(1,2,1)
    plt.plot(history_seg.history['loss'], label='train_loss')
    plt.plot(history_seg.history['val_loss'], label='val_loss')
    plt.title('Segmentation Loss'); plt.legend()
    plt.subplot(1,2,2)
    plt.plot(history_seg.history['accuracy'], label='train_acc')
    plt.plot(history_seg.history['val_accuracy'], label='val_acc')
    plt.title('Segmentation Accuracy'); plt.legend()
    plt.show()

#Testing & Metrics
y_true = test_gen.classes
y_pred_probs = clf_model.predict(test_gen, verbose=1)
y_pred = np.argmax(y_pred_probs, axis=1)

CONF_THRESH = 0.60
inv_map = {v: k for k, v in train_gen.class_indices.items()}
invalid_idx = train_gen.class_indices.get("Invalid Class", -1)

y_pred_adjusted = y_pred.copy()
for i, probs in enumerate(y_pred_probs):
    if np.max(probs) < CONF_THRESH and invalid_idx != -1:
        y_pred_adjusted[i] = invalid_idx

print("\n=== Classification Report ===")
print(classification_report(y_true, y_pred_adjusted, target_names=list(test_gen.class_indices.keys())))

# Show confusion matrix heatmap
cm = confusion_matrix(y_true, y_pred_adjusted)
plt.figure(figsize=(7,6))
sns.heatmap(cm, annot=True, fmt='d', xticklabels=list(test_gen.class_indices.keys()), yticklabels=list(test_gen.class_indices.keys()))
plt.xlabel('Predicted'); plt.ylabel('True'); plt.title('Confusion Matrix'); plt.show()

#Utility: severity and visualization helpers
def severity_from_mask_percent(mask_pred, threshold=0.5):
    # returns severity label and fraction in percent (0-100)
    bin_mask = (mask_pred > threshold).astype(np.float32)
    frac = bin_mask.sum() / bin_mask.size
    pct = float(frac * 100.0)
    # thresholds per your request: 0-15 mild, 15-30 moderate, >30 severe
    if pct <= 15.0:
        return "Mild", pct
    elif pct <= 30.0:
        return "Moderate", pct
    else:
        return "Severe", pct

def create_heatmap_overlay_rgb(original_rgb, mask_01):
    heatmap = cv2.applyColorMap(np.uint8(255 * mask_01), cv2.COLORMAP_JET)
    heatmap = cv2.cvtColor(heatmap, cv2.COLOR_BGR2RGB)
    heatmap = cv2.resize(heatmap, (original_rgb.shape[1], original_rgb.shape[0]))
    output = cv2.addWeighted(original_rgb, 0.6, heatmap, 0.4, 0)
    return output

def extract_bboxes_from_mask(mask_01_uint8, min_area=20):
    # mask_01_uint8: 0/1 uint8 at original image size
    contours, _ = cv2.findContours(mask_01_uint8, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    bboxes = []
    for cnt in contours:
        area = cv2.contourArea(cnt)
        if area < min_area:
            continue
        x,y,w,h = cv2.boundingRect(cnt)
        bboxes.append((x,y,w,h,area,cnt))
    # sort by area desc
    bboxes = sorted(bboxes, key=lambda x: x[4], reverse=True)
    return bboxes

#(test)visualization
filenames = test_gen.filenames
true_labels = test_gen.classes
class_idx_map = train_gen.class_indices  # mapping name->idx
inv_class_map = {v:k for k,v in class_idx_map.items()}

def show_examples_per_class(num_per_class=4):
    plt.rcParams.update({'font.size':10})
    for cls in classes:
        if cls not in class_idx_map:
            print("Class missing in train_gen:", cls); continue
        idx_cls = class_idx_map[cls]
        indices = np.where(true_labels == idx_cls)[0].tolist()
        if len(indices) == 0:
            print(f"No test examples for {cls}")
            continue
        selected = indices if len(indices) <= num_per_class else random.sample(indices, num_per_class)
        print(f"\nShowing {len(selected)} examples for class: {cls}")
        for i, idx in enumerate(selected):
            fname = filenames[idx]
            img_path = os.path.join(TEST_DIR, fname)
            img = cv2.imread(img_path)
            if img is None: continue
            img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
            pred_label = inv_map[y_pred_adjusted[idx]]
            conf = float(np.max(y_pred_probs[idx]))
            plt.figure(figsize=(5,4))
            plt.imshow(img_rgb)
            plt.title(f"{cls} | File:{os.path.basename(fname)}\nPred: {pred_label} ({conf:.2f})")
            plt.axis('off')
            plt.show()

show_examples_per_class(4)

#Malignant-specific detailed visualization
if seg_model is not None and "Malignant cases" in class_idx_map:
    mal_idx = class_idx_map["Malignant cases"]
    mal_indices = np.where(true_labels == mal_idx)[0].tolist()
    if len(mal_indices) == 0:
        print("No malignant test examples.")
    else:
        selected = mal_indices if len(mal_indices) <= 4 else random.sample(mal_indices, 4)
        print("\nDetailed malignant visualizations for", len(selected), "images.")
        for idx in selected:
            fname = filenames[idx]
            img_path = os.path.join(TEST_DIR, fname)
            img_bgr = cv2.imread(img_path)
            if img_bgr is None:
                continue
            img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
            # Prepare seg input
            seg_in = preprocess_seg_input(img_path, SEG_SIZE)
            if seg_in is None:
                continue
            pred_mask = seg_model.predict(np.expand_dims(seg_in,0), verbose=0)[0,:,:,0]
            # Resize mask to original image size for boxes/overlay
            mask_resized = cv2.resize((pred_mask*255).astype(np.uint8), (img_rgb.shape[1], img_rgb.shape[0]))
            mask_01_uint8 = (mask_resized > 127).astype(np.uint8)  # 0/1
            # Get bboxes
            bboxes = extract_bboxes_from_mask(mask_01_uint8, min_area=50)
            # compute severity (percent)
            _, pct = severity_from_mask_percent(pred_mask, threshold=0.5)
            sev_label, pct = severity_from_mask_percent(pred_mask, threshold=0.5)
            # heatmap overlay (on resized to original)
            mask_01_float_resized = cv2.resize((pred_mask).astype(np.float32), (img_rgb.shape[1], img_rgb.shape[0]))
            overlay = create_heatmap_overlay_rgb(img_rgb, mask_01_float_resized)
            # contour overlay
            contour_overlay = img_rgb.copy()
            contours, _ = cv2.findContours(mask_01_uint8, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
            cv2.drawContours(contour_overlay, contours, -1, (255,0,0), 2)  # blue-ish contour
            # bbox overlay
            bbox_overlay = img_rgb.copy()
            for (x,y,w,h,area,cnt) in bboxes[:5]:  # draw up to 5 boxes
                cv2.rectangle(bbox_overlay, (x,y), (x+w, y+h), (255,255,0), 2)  # cyan box
                cv2.putText(bbox_overlay, f"{area:.0f}", (x, max(y-6,0)), cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255,255,0), 2)
            # binary mask display resized to original shape
            mask_display = mask_01_uint8 * 255

            # Plot 1: Original
            plt.figure(figsize=(16,6))
            plt.subplot(1,4,1)
            plt.imshow(img_rgb)
            plt.title(f"Original\nFile: {os.path.basename(fname)}")
            plt.axis('off')
            # Plot 2: Binary mask (full size)
            plt.subplot(1,4,2)
            plt.imshow(mask_display, cmap='gray')
            plt.title(f"Binary Mask\nSeverity: {sev_label} ({pct:.1f}%)")
            plt.axis('off')
            # Plot 3: Heatmap overlay
            plt.subplot(1,4,3)
            plt.imshow(overlay)
            plt.title("Heatmap Overlay (Seg mask)")
            plt.axis('off')
            # Plot 4: BBoxes + contours
            plt.subplot(1,4,4)
            plt.imshow(bbox_overlay)
            plt.title("Bounding Boxes + Area (px)")
            plt.axis('off')
            plt.show()
else:
    print("Segmentation model not available - skipping malignant detailed visualization.")

#Save evaluation outputs / example metadata
# Save a small JSON report of a few metrics & example severities
report = {
    "classification_report": classification_report(y_true, y_pred_adjusted, target_names=list(test_gen.class_indices.keys()), output_dict=True),
    "confusion_matrix": cm.tolist()
}
with open(os.path.join(RESULTS_DIR, "evaluation_report.json"), "w") as f:
    json.dump(report, f, indent=2)

#TFLite Conversion for Flutter
print("\nConverting to TFLite... (classification)")
# ensure saved Keras model loaded
clf_saved = tf.keras.models.load_model(os.path.join(RESULTS_DIR, "final_classification_model.h5"))
converter = tf.lite.TFLiteConverter.from_keras_model(clf_saved)
converter.optimizations = [tf.lite.Optimize.DEFAULT]
tflite_clf = converter.convert()
with open(os.path.join(TFLITE_DIR, "classification_model.tflite"), "wb") as f:
    f.write(tflite_clf)
print("Saved:", os.path.join(TFLITE_DIR, "classification_model.tflite"))

if seg_model is not None:
    print("Converting segmentation model to TFLite...")
    seg_saved = tf.keras.models.load_model(os.path.join(RESULTS_DIR, "final_segmentation_model.h5"))
    converter = tf.lite.TFLiteConverter.from_keras_model(seg_saved)
    converter.optimizations = [tf.lite.Optimize.DEFAULT]
    tflite_seg = converter.convert()
    with open(os.path.join(TFLITE_DIR, "segmentation_model.tflite"), "wb") as f:
        f.write(tflite_seg)
    print("Saved:", os.path.join(TFLITE_DIR, "segmentation_model.tflite"))
else:
    print("Segmentation model missing - no tflite segmentation saved.")

print("\n✅ Full pipeline finished. Models + visuals saved in:", RESULTS_DIR)
```
